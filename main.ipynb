```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 1. Génération de données synthétiques (bruitées autour d'une droite)
np.random.seed(42)
X = np.linspace(0, 10, 50)
# Équation sous-jacente : y = 2.5 * x + 5 + bruit aléatoire
Y = 2.5 * X + 5 + np.random.normal(0, 2, size=X.shape)

# 2. Calcul de la régression linéaire (degré 1)
# np.polyfit renvoie les coefficients [pente (m), ordonnée à l'origine (c)]
m, c = np.polyfit(X, Y, 1)
Y_pred = m * X + c

# 3. Visualisation avec Matplotlib
plt.figure(figsize=(8, 5))

# Nuage de points (données d'origine)
plt.scatter(X, Y, color='blue', label='Données observées', alpha=0.7)

# Habillage du graphique
plt.title('Présentation d\'une Régression Linéaire Basique', fontsize=12, fontweight='bold')
plt.xlabel('Variable Indépendante (X)')
plt.ylabel('Variable Dépendante (Y)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Affichage
plt.show()
```

```python
# 1. Génération de données synthétiques (bruitées autour d'une droite)
np.random.seed(42)
X = np.linspace(0, 10, 50)
# Équation sous-jacente : y = 2.5 * x + 5 + bruit aléatoire
Y = 2.5 * X + 5 + np.random.normal(0, 2, size=X.shape)

# 2. Calcul de la régression linéaire (degré 1)
# np.polyfit renvoie les coefficients [pente (m), ordonnée à l'origine (c)]
m, c = np.polyfit(X, Y, 1)
Y_pred = m * X + c

# 3. Visualisation avec Matplotlib
plt.figure(figsize=(8, 5))

# Nuage de points (données d'origine)
plt.scatter(X, Y, color='blue', label='Données observées', alpha=0.7)

# Ligne de régression
plt.plot(X, Y_pred, color='red', linewidth=2, label=f'Régression : y = {m:.2f}x + {c:.2f}')

# Habillage du graphique
plt.title('Présentation d\'une Régression Linéaire Basique', fontsize=12, fontweight='bold')
plt.xlabel('Variable Indépendante (X)')
plt.ylabel('Variable Dépendante (Y)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Affichage
plt.show()
```
